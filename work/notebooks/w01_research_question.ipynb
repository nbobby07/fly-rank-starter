## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
trend = (
    df["trend_direction"]
    .astype("string")
    .str.strip()
    .str.lower()
)

downward_pages = trend.eq("down")

total_pages = len(df)
total_clients = df["client_id"].nunique()
downward_count = int(downward_pages.sum())
downward_percentage = downward_pages.mean()

print(f"Total content pages: {total_pages:,}")
print(f"Total pseudonymized clients: {total_clients:,}")
print(f"Pages with an observed downward trend: {downward_count:,}")
print(f"Percentage with an observed downward trend: {downward_percentage:.1%}")
```
